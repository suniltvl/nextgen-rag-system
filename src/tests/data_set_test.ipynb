### Data Loader

#### Create Loader

```python
try:
    def get_loader():
        """
        Data loading component of the RAG system. This function initializes the data loader based on the configuration and loads the data.
        """
        
        logger.info("Data loader initialized")
        loader = DataLoaderFactory.create(config.data_loader)
        return loader
    print("✓ get_loader function defined successfully")
except Exception as e:
    print(f"✗ Error defining get_loader function: {str(e)}")
    raise
```

#### Loading Data

```python
try:
    def get_data(loader=None):
        
        if(loader is None):
            loader = get_loader()
            
        ds = loader.load() 
        logger.info(f"Data loaded: {len(ds)} documents found")
        return ds["documents"]
    print("✓ get_data function defined successfully")
except Exception as e:
    print(f"✗ Error defining get_data function: {str(e)}")
    raise
```

```python
try:
    ds = get_data()
    print(f"✓ Dataset loaded successfully")
    print(f"Dataset Count: {len(ds)}")
except Exception as e:
    print(f"✗ Error loading dataset: {str(e)}")
    raise
```

### Generator

#### Create Generator

```python
try:
    generator_config = config.generator
    generator_config.api_key = "lm-studio"
    generator_config.base_url = "http://localhost:1234/v1"

    generator = GeneratorFactory.create(generator_config)
    print("✓ Generator created successfully")
    print(f"Generator config: {generator_config}")
except Exception as e:
    print(f"✗ Error creating generator: {str(e)}")
    raise
```

#### Generate Response

```python
try:
    query = "give me embedding model?"
    context = retrieved_data
    system_prompt = BASIC_RAG_PROMPT

    answer = generator.generate(
        question=query,
        context=context,
        system_prompt=system_prompt
    )

    print("✓ Response generated successfully")
    print(f"Query: {query}")
    print(f"Answer: {answer}")
except Exception as e:
    print(f"✗ Error generating response: {str(e)}")
    raise
```
