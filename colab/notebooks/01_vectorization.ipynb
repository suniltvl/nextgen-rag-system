Load ragbench parquet from Drive, embed with local model, build/load Chroma index on Drive.

```python
DRIVE_ROOT = "/content/drive/MyDrive/capstone-rag-kag"
PRESET = "covidqa_tuned"
DOMAIN = "biomedical"
DATASET = None
REBUILD_INDEX = False
```

```python
import sys
from pathlib import Path

try:
    from google.colab import drive
    drive.mount("/content/drive")
except ImportError:
    pass

REPO_ROOT = Path(DRIVE_ROOT) / "repo"
sys.path.insert(0, str(REPO_ROOT))

from colab.lib.colab_setup import bootstrap, print_experiment
from colab.lib.runner import build_index

exp = bootstrap(
    drive_root=DRIVE_ROOT,
    preset=PRESET,
    domain=DOMAIN,
    dataset=DATASET,
    overrides={"rebuild_index": REBUILD_INDEX},
    install_deps=True,
    mount=False,
)
print_experiment(exp)
```

```python
retriever = build_index(exp)
print(f"Chroma ready at: {exp.chroma_dir}")
```
