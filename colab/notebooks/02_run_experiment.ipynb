RAG generation + TRACe judge with OpenAI. Results persist to SQLite on Drive.

Set `samples=2` in overrides for a quick smoke test.

```python
DRIVE_ROOT = "/content/drive/MyDrive/capstone-rag-kag"
PRESET = "covidqa_tuned"
DOMAIN = "biomedical"
DATASET = None

OVERRIDES = {
    "samples": 50,
    "max_workers": 2,
    "rebuild_db": False,
    "rebuild_index": False,
    "resume": True,
}
```

```python
import sys
from pathlib import Path

try:
    from google.colab import drive
    drive.mount("/content/drive")
except ImportError:
    pass

REPO_ROOT = Path(DRIVE_ROOT) / "repo"
sys.path.insert(0, str(REPO_ROOT))

from colab.lib.colab_setup import bootstrap, print_experiment
from colab.lib.runner import run_experiment

exp = bootstrap(
    drive_root=DRIVE_ROOT,
    preset=PRESET,
    domain=DOMAIN,
    dataset=DATASET,
    overrides=OVERRIDES,
    install_deps=True,
    mount=False,
)
print_experiment(exp)
```
